```python
import pandas as pd
import numpy as np

# Cargar la base de datos
ruta_base = r"C:\Users\Dafne\Documents\GitHub\Base-de-datos-investigacion-Economica\Base24_25_v2\BASE_REGRESIONES.csv"
df = pd.read_csv(ruta_base)

print("="*80)
print("REVISIÓN COMPLETA DE LA BASE DE DATOS")
print("="*80)

# 1. INFORMACIÓN GENERAL
print("\n1. INFORMACIÓN GENERAL")
print("-"*40)
print(f"Número de observaciones (filas): {len(df)}")
print(f"Número de variables (columnas): {len(df.columns)}")
print(f"Memoria utilizada: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

# 2. LISTA COMPLETA DE VARIABLES
print("\n2. LISTA COMPLETA DE VARIABLES")
print("-"*40)
for i, columna in enumerate(df.columns, 1):
    print(f"{i:3}. {columna}")

# 3. TIPO DE DATOS DE CADA VARIABLE
print("\n3. TIPO DE DATOS")
print("-"*40)
tipos = df.dtypes
for columna, tipo in tipos.items():
    print(f"{columna:30} -> {tipo}")

# 4. RESULTADOS DE NULOS Y VALORES ÚNICOS
print("\n4. ANÁLISIS DE VALORES FALTANTES Y ÚNICOS")
print("-"*40)
print(f"{'Variable':30} {'Nulos':>8} {'% Nulos':>10} {'Valores Únicos':>15}")
print("-"*70)
for columna in df.columns:
    nulos = df[columna].isna().sum()
    pct_nulos = (nulos / len(df)) * 100
    unicos = df[columna].nunique()
    print(f"{columna:30} {nulos:>8} {pct_nulos:>9.1f}% {unicos:>15}")

# 5. ESTADÍSTICAS DESCRIPTIVAS PARA VARIABLES NUMÉRICAS
print("\n5. ESTADÍSTICAS DESCRIPTIVAS - VARIABLES NUMÉRICAS")
print("-"*40)
num_cols = df.select_dtypes(include=[np.number]).columns
if len(num_cols) > 0:
    print(df[num_cols].describe())
else:
    print("No hay variables numéricas")

# 6. VARIABLES CATEGÓRICAS
print("\n6. VARIABLES CATEGÓRICAS (TEXTO/OBJETO)")
print("-"*40)
cat_cols = df.select_dtypes(include=['object']).columns
if len(cat_cols) > 0:
    for col in cat_cols:
        print(f"\n{col}:")
        print(df[col].value_counts().head(10))
else:
    print("No hay variables categóricas")

# 7. ANÁLISIS DE CORRELACIÓN (solo numéricas)
print("\n7. MATRIZ DE CORRELACIÓN (principales variables numéricas)")
print("-"*40)
if len(num_cols) > 1:
    # Mostrar top correlaciones
    corr_matrix = df[num_cols].corr()
    # Crear lista de correlaciones altas
    high_corr = []
    for i in range(len(corr_matrix.columns)):
        for j in range(i+1, len(corr_matrix.columns)):
            corr_val = corr_matrix.iloc[i, j]
            if abs(corr_val) > 0.5:
                high_corr.append((corr_matrix.columns[i], corr_matrix.columns[j], corr_val))
    
    if high_corr:
        print("Correlaciones altas (>|0.5|):")
        for var1, var2, corr in sorted(high_corr, key=lambda x: abs(x[2]), reverse=True):
            print(f"  {var1} <-> {var2}: {corr:.3f}")
    else:
        print("No hay correlaciones altas (>|0.5|)")
else:
    print("Se necesitan al menos 2 variables numéricas para correlación")

# 8. MUESTRA DE DATOS
print("\n8. MUESTRA DE DATOS (primeras 5 filas)")
print("-"*40)
print(df.head())

# 9. MUESTRA ALEATORIA
print("\n9. MUESTRA ALEATORIA (5 filas aleatorias)")
print("-"*40)
print(df.sample(min(5, len(df))))

# 10. IDENTIFICACIÓN DE VARIABLES POTENCIALES PARA REGRESIÓN
print("\n10. SUGERENCIAS PARA REGRESIÓN")
print("-"*40)

# Identificar variable dependiente potencial (variable numérica con nombre que sugiere ser objetivo)
potential_y = []
for col in num_cols:
    if any(keyword in col.lower() for keyword in ['precio', 'venta', 'ingreso', 'ganancia', 'produccion', 'pib', 'empleo', 'salario', 'educacion']):
        potential_y.append(col)

if potential_y:
    print("Posibles variables dependientes (Y):")
    for col in potential_y:
        print(f"  - {col}")
else:
    print("No se identificaron variables dependientes obvias")

# Variables explicativas potenciales (numéricas que no son la dependiente)
print("\nPosibles variables independientes (X):")
for col in num_cols:
    if col not in potential_y:
        print(f"  - {col}")

# 11. CHECK DE DUPLICADOS
print("\n11. DATOS DUPLICADOS")
print("-"*40)
duplicados = df.duplicated().sum()
print(f"Filas duplicadas: {duplicados}")
if duplicados > 0:
    print(f"Porcentaje de duplicados: {(duplicados/len(df))*100:.2f}%")

# 12. RANGO DE FECHAS (si hay columnas de fecha)
print("\n12. COLUMNAS DE FECHA (si existen)")
print("-"*40)
date_cols = df.select_dtypes(include=['datetime64']).columns
if len(date_cols) > 0:
    for col in date_cols:
        print(f"{col}: {df[col].min()} a {df[col].max()}")
else:
    # Intentar identificar columnas que parecen fechas
    potential_dates = []
    for col in df.columns:
        if any(keyword in col.lower() for keyword in ['fecha', 'date', 'año', 'year', 'mes', 'month', 'dia', 'day']):
            potential_dates.append(col)
    if potential_dates:
        print("Posibles columnas de fecha:")
        for col in potential_dates:
            print(f"  - {col} (tipo: {df[col].dtype})")
    else:
        print("No se encontraron columnas de fecha")

print("\n" + "="*80)
print("FIN DE LA REVISIÓN")
print("="*80)
```
