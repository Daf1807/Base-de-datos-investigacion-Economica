```python
import pandas as pd
import os

# ============================================================
# 1. RUTAS DE LOS ARCHIVOS ORIGINALES DEL INEI (2024)
# ============================================================
base = r"C:\Users\Dafne\Downloads\Base Datos"

modulos = {
    "Mod100": (os.path.join(base, "966-Modulo01"), "Enaho01-2024-100"),
    "Mod200": (os.path.join(base, "966-Modulo02"), "Enaho01-2024-200"),
    "Mod300": (os.path.join(base, "966-Modulo03"), "Enaho01A-2024-300"),
    "Mod500": (os.path.join(base, "966-Modulo05"), "Enaho01a-2024-500"),
    "Sumaria": (os.path.join(base, "966-Modulo34"), "Sumaria-2024-12g")
}

# ============================================================
# 2. FUNCIONES INTELIGENTES DE LECTURA
# ============================================================
def encontrar_archivo(carpeta, inicio_nombre):
    for f in os.listdir(carpeta):
        if f.upper().replace(".CSV", "").startswith(inicio_nombre.upper()):
            return os.path.join(carpeta, f)
    raise FileNotFoundError(f"No se encontró {inicio_nombre} en {carpeta}")

def leer_csv_inei(ruta):
    for sep in [",", ";"]:
        for enc in ["utf-8-sig", "latin1"]:
            try:
                df = pd.read_csv(ruta, sep=sep, encoding=enc, low_memory=False, dtype=str)
                if len(df.columns) > 1:
                    df.columns = df.columns.str.strip()
                    return df
            except Exception:
                pass
    raise ValueError(f"No se pudo leer el archivo: {ruta}")

# ============================================================
# 3. EXTRACCIÓN DE VARIABLES (TABLA A DE TU TESIS)
# ============================================================
print("Cargando módulos y extrayendo variables...")

ruta_200 = encontrar_archivo(modulos["Mod200"][0], modulos["Mod200"][1])
df200 = leer_csv_inei(ruta_200)[["CONGLOME", "VIVIENDA", "HOGAR", "CODPERSO", "P203", "P207", "P208A"]]

ruta_300 = encontrar_archivo(modulos["Mod300"][0], modulos["Mod300"][1])
df300 = leer_csv_inei(ruta_300)[["CONGLOME", "VIVIENDA", "HOGAR", "CODPERSO", "P301A", "P301B"]]

ruta_500 = encontrar_archivo(modulos["Mod500"][0], modulos["Mod500"][1])
df500 = leer_csv_inei(ruta_500)[["CONGLOME", "VIVIENDA", "HOGAR", "CODPERSO", 
                                "P507", "P510A1", "P511A", "P512A", 
                                "P558E1_4", "P558E1_8", "P558E1_9"]]

ruta_100 = encontrar_archivo(modulos["Mod100"][0], modulos["Mod100"][1])
df100 = leer_csv_inei(ruta_100)
cols_100 = ["CONGLOME", "VIVIENDA", "HOGAR", "DOMINIO", "ESTRATO", "PANEL"]
for fac in ["FACTOR", "FACPOB", "FACTOR07"]:
    if fac in df100.columns:
        cols_100.append(fac)
        break
df100 = df100[cols_100]

ruta_sum = encontrar_archivo(modulos["Sumaria"][0], modulos["Sumaria"][1])
df_sum = leer_csv_inei(ruta_sum)
cols_sum = ["CONGLOME", "VIVIENDA", "HOGAR"]
for g in ["GASHOG2", "GASHOG1"]:
    if g in df_sum.columns:
        cols_sum.append(g); break
for m in ["MIEPERHO", "MIEMBRO"]:
    if m in df_sum.columns:
        cols_sum.append(m); break
df_sum = df_sum[cols_sum]

print("✅ Extracción lista. Uniendo bases...")

# ============================================================
# 4. UNIR TODO EN UN SOLO DATAFRAME
# ============================================================
llave_persona = ["CONGLOME", "VIVIENDA", "HOGAR", "CODPERSO"]
llave_hogar = ["CONGLOME", "VIVIENDA", "HOGAR"]

df = df200.merge(df300, on=llave_persona, how="left")
df = df.merge(df500, on=llave_persona, how="left")
df = df.merge(df100, on=llave_hogar, how="left")
df = df.merge(df_sum, on=llave_hogar, how="left")

for c in llave_persona:
    df[c] = df[c].astype(str).str.strip().str.replace(r'\.0$', '', regex=True)

df["llave_persona"] = df[llave_persona].agg("-".join, axis=1)

# ============================================================
# 5. CONSTRUIR VARIABLES PARA LA TESIS (CORREGIDO)
# ============================================================
print("Construyendo variables de investigación...")

# Función segura para convertir texto a números (maneja espacios vacíos del INEI)
def to_num(serie):
    return pd.to_numeric(serie, errors="coerce")

# Edad numérica
df["Edad"] = to_num(df["P208A"])

# Ocupado (1 si P507 no es vacío)
df["Ocupado"] = df["P507"].notna().astype(int)

# Informalidad (Criterio INEI: Sector + Protección)
df["P510A1_num"] = to_num(df["P510A1"])
df["P511A_num"] = to_num(df["P511A"])
df["P507_num"] = to_num(df["P507"])

tfnr = df["P507_num"] == 5
trab_hogar = df["P507_num"] == 6
sin_sunat = df["P510A1_num"] == 3
dependiente = df["P507_num"].isin([3, 4])
sin_contrato = df["P511A_num"] == 7

df["Informal"] = (
    sin_sunat.fillna(False) | 
    tfnr.fillna(False) | 
    trab_hogar.fillna(False) | 
    (dependiente & sin_contrato.fillna(False))
).astype(int)
df.loc[df["Ocupado"] == 0, "Informal"] = pd.NA

# CRÉDITO Y BILLETERA (Manejando espacios vacíos con to_num)
df["P558E1_4"] = to_num(df["P558E1_4"]).fillna(0)
df["P558E1_9"] = to_num(df["P558E1_9"]).fillna(0)
df["P558E1_8"] = to_num(df["P558E1_8"]).fillna(0)

df["CreditoFormal"] = ((df["P558E1_4"] == 4) | (df["P558E1_9"] == 9)).astype(int)
df["Billetera"] = (df["P558E1_8"] == 8).astype(int)

# Gasto per cápita
if "GASHOG2" in df.columns and "MIEPERHO" in df.columns:
    df["Gasto_per_capita"] = to_num(df["GASHOG2"]) / to_num(df["MIEPERHO"])

# ============================================================
# 6. FILTROS Y GUARDADO
# ============================================================
print("\n" + "=" * 50)
print("DIAGNÓSTICO DE LA MUESTRA")
print("=" * 50)

n_total = len(df)
print(f"Total personas encuestadas: {n_total:,}")

df_mayor = df[df["Edad"] >= 14]
print(f"Mayores de 14 años: {len(df_mayor):,}")

df_ocup = df_mayor[df_mayor["Ocupado"] == 1]
print(f"Ocupados: {len(df_ocup):,}")

# Guardar el archivo maestro
ruta_salida = os.path.join(base, "MASTER_2024.csv")
df.to_csv(ruta_salida, index=False, encoding="utf-8-sig")

print("\n" + "=" * 50)
print(f"✅ ARCHIVO MAESTRO GUARDADO EN:\n{ruta_salida}")
print(f"Tamaño final: {df.shape[0]:,} filas x {df.shape[1]:,} columnas")
```

```python
import pandas as pd
import os
from scipy import stats

base = r"C:\Users\Dafne\Downloads\Base Datos"

# 1. CARGAR BASES
df_2024 = pd.read_csv(os.path.join(base, "MASTER_2024.csv"), encoding="utf-8-sig", low_memory=False)
df_2025 = pd.read_csv(os.path.join(base, "MASTER_2025.csv"), encoding="utf-8-sig", low_memory=False)

# 2. LIMPIAR LLAVES
llaves = ["CONGLOME", "VIVIENDA", "HOGAR", "CODPERSO"]
for df in [df_2024, df_2025]:
    for c in llaves:
        df[c] = df[c].astype(str).str.strip().str.replace(r'\.0$', '', regex=True)
    df["llave_persona"] = df[llaves].agg("-".join, axis=1)

# 3. IDENTIFICAR ATRICIÓN
# 1 = Permanente (está en ambos años), 0 = Atricionado (solo está en 2024)
df_2024["Permanente"] = df_2024["llave_persona"].isin(df_2025["llave_persona"]).astype(int)

# 4. PREPARAR VARIABLES PARA COMPARAR
df_2024["Mujer"] = (df_2024["P207"] == 2).astype(int)
df_2024["Informal_num"] = pd.to_numeric(df_2024["Informal"], errors="coerce")

# Variables a comparar (las relevantes para tu tesis)
vars_comparar = ["Edad", "Mujer", "Billetera", "CreditoFormal", "Ocupado", "Informal_num"]

# 5. TABLA DE MEDIAS Y PRUEBA T
print("=" * 75)
print("TABLA DE ATRICIÓN: COMPARACIÓN DE MEDIAS 2024 (Sección 2.2.4)")
print("=" * 75)
print(f"{'Variable':<15} | {'Permanentes':>12} | {'Atricionados':>13} | {'Dif.':>8} | {'Valor p':>8}")
print("-" * 75)

n_perm = df_2024["Permanente"].sum()
n_atrit = len(df_2024) - n_perm
print(f"{'N (Personas)':<15} | {n_perm:>12,} | {n_atrit:>13,} | {'':>8} | {'':>8}")
print("-" * 75)

resultados = []
for var in vars_comparar:
    # Separar grupos y quitar vacíos
    grupo_perm = df_2024.loc[df_2024["Permanente"] == 1, var].dropna()
    grupo_atrit = df_2024.loc[df_2024["Permanente"] == 0, var].dropna()
    
    # Medias
    media_perm = grupo_perm.mean()
    media_atrit = grupo_atrit.mean()
    diff = media_perm - media_atrit
    
    # Prueba t (Welch's t-test para varianzas desiguales)
    t_stat, p_val = stats.ttest_ind(grupo_perm, grupo_atrit, equal_var=False)
    
    # Significancia
    sig = "***" if p_val < 0.01 else ("**" if p_val < 0.05 else ("*" if p_val < 0.10 else ""))
    
    print(f"{var:<15} | {media_perm:>12.4f} | {media_atrit:>13.4f} | {diff:>+8.4f} | {p_val:>8.4f} {sig}")
    
    resultados.append((var, media_perm, media_atrit, diff, p_val))

print("=" * 75)
print("Notas: *** p<0.01, ** p<0.05, * p<0.10. Dif. = Permanentes - Atricionados")
print("Informal_num solo se calcula para ocupados, por eso el N interno es menor.")
```

```python
2.5. Potencia estadística y efecto mínimo detectable
```

```python
import numpy as np

# ==========================================
# Parámetros de Potencia (80% y 5% significancia)
# ==========================================
z_alpha = 1.96  # 5% two-sided
z_beta = 0.84   # 80% power
z_sum = z_alpha + z_beta  # Aproximadamente 2.80

# Datos de las 4 celdas (Tasa base y N)
# Celda 1: Informal, Billetera=1
p1, n1 = 0.1383, 933
# Celda 2: Informal, Billetera=0
p2, n2 = 0.0506, 4324
# Celda 3: Formal, Billetera=1
p3, n3 = 0.2264, 773
# Celda 4: Formal, Billetera=0
p4, n4 = 0.0955, 1592

# Varianzas de cada celda: p(1-p)/n
v1 = p1 * (1 - p1) / n1
v2 = p2 * (1 - p2) / n2
v3 = p3 * (1 - p3) / n3
v4 = p4 * (1 - p4) / n4

# EMD para el efecto de Billetera dentro de cada grupo (diferencia simple)
emd_inf = z_sum * np.sqrt(v1 + v2)
emd_for = z_sum * np.sqrt(v3 + v4)

# EMD para la INTERACCIÓN (diferencia de diferencias)
# La varianza de la interacción es la suma de las varianzas de las 4 celdas
emd_interaccion = z_sum * np.sqrt(v1 + v2 + v3 + v4)

print("=" * 70)
print("TABLA DE POTENCIA ESTADÍSTICA (Sección 2.5)")
print("=" * 70)
print(f"{'Celda':<25} | {'N':>6} | {'Tasa base':>10} | {'EMD (p.p.)':>10}")
print("-" * 70)
print(f"{'Informal, billetera = 1':<25} | {n1:>6,} | {p1*100:>9.2f}% | {'':>10}")
print(f"{'Informal, billetera = 0':<25} | {n2:>6,} | {p2*100:>9.2f}% | {emd_inf*100:>9.2f}")
print(f"{'Formal, billetera = 1':<25} | {n3:>6,} | {p3*100:>9.2f}% | {'':>10}")
print(f"{'Formal, billetera = 0':<25} | {n4:>6,} | {p4*100:>9.2f}% | {emd_for*100:>9.2f}")
print("-" * 70)
print(f"{'Interacción (H2)':<25} | {'':>6} | {'':>10} | {emd_interaccion*100:>9.2f}")
print("=" * 70)
print(f"\nNota: EMD calculado con 80% de potencia y 5% de nivel de significancia.")
print(f"El diseño puede detectar interacciones de {emd_interaccion*100:.2f} p.p. o más.")
```

```python
import pandas as pd
import statsmodels.api as sm
import os

base = r"C:\Users\Dafne\Downloads\Base Datos"
df = pd.read_csv(os.path.join(base, "BASE_REGRESIONES.csv"), encoding="utf-8-sig", low_memory=False)

# 1. PREPARAR VARIABLES
df["Informal_num"] = pd.to_numeric(df["Informal"], errors="coerce")
df["Edad_num"] = pd.to_numeric(df["Edad"], errors="coerce")
df["P301A_num"] = pd.to_numeric(df["P301A"], errors="coerce")
df["MIEPERHO_num"] = pd.to_numeric(df["MIEPERHO"], errors="coerce")

# Dummies y términos
df["Mujer"] = (df["P207"] == 2).astype(int)
df["Urbano"] = (pd.to_numeric(df["ESTRATO"], errors="coerce") <= 6).astype(int)
df["Edad_cuadrado"] = df["Edad_num"]**2
df["Interaccion"] = df["Billetera"] * df["Informal_num"]

# 2. LIMPIAR VACÍOS PARA REGRESIÓN
vars_reg = ["NuevoCredito", "Billetera", "Informal_num", "Interaccion", 
            "Edad_num", "Edad_cuadrado", "Mujer", "P301A_num", "Urbano", 
            "MIEPERHO_num", "DOMINIO"]

df_reg = df.dropna(subset=vars_reg).copy()

# 3. DEFINIR Y Y X
Y = df_reg["NuevoCredito"]

# Variables continuas y dummies
X = df_reg[["Billetera", "Informal_num", "Interaccion", "Edad_num", 
            "Edad_cuadrado", "Mujer", "P301A_num", "Urbano", "MIEPERHO_num"]]

# Agregar efectos fijos de Dominio (categorical)
X = pd.concat([X, pd.get_dummies(df_reg["DOMINIO"], prefix="DOM", drop_first=True, dtype=float)], axis=1)

# Agregar constante
X = sm.add_constant(X)

# 4. CORRER EL MODELO PROBIT
print("Corriendo modelo Probit...")
modelo = sm.Probit(Y, X).fit()
print("\n" + "=" * 60)
print("RESULTADOS DEL PROBIT (Coeficientes base)")
print("=" * 60)
print(modelo.summary())

# 5. CALCULAR EFECTOS MARGINALES PROMEDIO (EMP) - Cumplimiento [P1]
print("\n" + "=" * 60)
print("EFECTOS MARGINALES PROMEDIO (EMP) - TABLA D")
print("=" * 60)
marg_eff = modelo.get_margeff(at='overall')
print(marg_eff.summary())

# Extraer valores para la tabla
print("\n" + "=" * 60)
print("RESUMEN DE EFECTOS MARGINALES (Para copiar a Word)")
print("=" * 60)
df_marg = pd.DataFrame({
    'EMP (puntos porcentuales)': marg_eff.margeff * 100,
    'Error estandar': marg_eff.margeff_se,
    'Valor p': marg_eff.margeff_pvalues
})
print(df_marg.head(4)) # Mostramos las 4 principales variables
```

```python
import pandas as pd
import os

base = r"C:\Users\Dafne\Downloads\Base Datos"

# 1. Cargar base final
df_reg = pd.read_csv(os.path.join(base, "BASE_REGRESIONES.csv"), encoding="utf-8-sig", low_memory=False)

# Cargar crudo 2025
carpeta_500_2025 = os.path.join(base, "1031-Modulo05-2025")
archivo_500_25 = [f for f in os.listdir(carpeta_500_2025) if f.upper().startswith("ENAHO01A-2025-500")][0]
ruta_500_25 = os.path.join(carpeta_500_2025, archivo_500_25)

# Leer seguro
df_raw = None
for sep in [",", ";"]:
    for enc in ["utf-8-sig", "latin1"]:
        try:
            temp = pd.read_csv(ruta_500_25, sep=sep, encoding=enc, low_memory=False, dtype=str)
            if len(temp.columns) > 1:
                df_raw = temp
                break
        except:
            pass
    if df_raw is not None:
        break

# 2. Normalizar llaves
def normalizar_llave(serie):
    s = serie.astype(str).str.strip()
    s = s.replace({"nan": None, "None": None, "NaN": None, "": None})
    s = s.str.replace(r"\.0+$", "", regex=True)
    num = pd.to_numeric(s, errors="coerce")
    if num.notna().sum() == s.notna().sum():
        return num.astype("Int64").astype(str)
    return s.str.lstrip("0").replace("", "0")

llaves = ["CONGLOME", "VIVIENDA", "HOGAR", "CODPERSO"]
for c in llaves:
    df_raw[f"_k_{c}"] = normalizar_llave(df_raw[c])
    df_reg[f"_k_{c}"] = normalizar_llave(df_reg[c])
    
df_raw["_key"] = df_raw[["_k_"+c for c in llaves]].agg("-".join, axis=1)
df_reg["_key"] = df_reg[["_k_"+c for c in llaves]].agg("-".join, axis=1)

# 3. Extraer CRÉDITO INFORMAL CORRECTO: P558G7 (valor 7 = Préstamo de prestamista/familiares)
if "P558G7" in df_raw.columns:
    print("✅ Variable correcta encontrada: P558G7")
    df_raw["CreditoInformal_2025"] = (pd.to_numeric(df_raw["P558G7"], errors="coerce").fillna(0) == 7).astype(int)
else:
    print("❌ No se encontró P558G7")
    
# 4. Hacer el Merge
df_extra = df_raw[["_key", "CreditoInformal_2025"]].drop_duplicates(subset="_key")

# Borrar la columna vieja que tenía el error
if "CreditoInformal_2025" in df_reg.columns:
    df_reg = df_reg.drop(columns=["CreditoInformal_2025"])
    
df_reg = df_reg.merge(df_extra, on="_key", how="left")

# Limpiar columnas auxiliares
df_reg = df_reg.drop(columns=["_key"] + ["_k_"+c for c in llaves])

# 5. Verificar y Guardar
print("\n" + "=" * 60)
print("VERIFICACIÓN DEL CRÉDITO INFORMAL (CORREGIDO)")
print("=" * 60)
print(df_reg["CreditoInformal_2025"].value_counts(dropna=False))

ruta_salida = os.path.join(base, "BASE_REGRESIONES.csv")
df_reg.to_csv(ruta_salida, index=False, encoding="utf-8-sig")
print("\n✅ Base corregida guardada. Ya tiene la variable correcta.")
```

```python
import pandas as pd
import statsmodels.api as sm
import os

base = r"C:\Users\Dafne\Downloads\Base Datos"
df = pd.read_csv(os.path.join(base, "BASE_REGRESIONES.csv"), encoding="utf-8-sig", low_memory=False)

# 1. Filtrar SOLO informales
df_inf = df[df["Informal_num"] == 1].copy()

# 2. Crear variable Ambos_2025 (1 si tiene crédito formal Y crédito informal en 2025)
df_inf["CreditoFormal_2025_num"] = pd.to_numeric(df_inf["CreditoFormal_2025"], errors="coerce")
df_inf["CreditoInformal_2025_num"] = pd.to_numeric(df_inf["CreditoInformal_2025"], errors="coerce")
df_inf["Ambos_2025"] = ((df_inf["CreditoFormal_2025_num"] == 1) & (df_inf["CreditoInformal_2025_num"] == 1)).astype(int)

# 3. Preparar controles
df_inf["Edad_num"] = pd.to_numeric(df_inf["Edad"], errors="coerce")
df_inf["P301A_num"] = pd.to_numeric(df_inf["P301A"], errors="coerce")
df_inf["MIEPERHO_num"] = pd.to_numeric(df_inf["MIEPERHO"], errors="coerce")
df_inf["Edad_cuadrado"] = df_inf["Edad_num"]**2
df_inf["Mujer"] = (df_inf["P207"] == 2).astype(int)
df_inf["Urbano"] = (pd.to_numeric(df_inf["ESTRATO"], errors="coerce") <= 6).astype(int)

# 4. Definir Y y X
vars_reg = ["Ambos_2025", "Billetera", "Edad_num", "Edad_cuadrado", "Mujer", "P301A_num", "Urbano", "MIEPERHO_num"]
df_reg = df_inf.dropna(subset=vars_reg).copy()

Y = df_reg["Ambos_2025"]
X = df_reg[["Billetera", "Edad_num", "Edad_cuadrado", "Mujer", "P301A_num", "Urbano", "MIEPERHO_num"]]
X = sm.add_constant(X)

# 5. Correr Probit
print("Corriendo Probit para 2.6.5 (Coexistencia de fuentes)...")
modelo_co = sm.Probit(Y, X).fit()
marg_co = modelo_co.get_margeff(at='overall')

print("\n" + "=" * 60)
print("EFECTOS MARGINALES COEXISTENCIA (Ambos_2025)")
print("=" * 60)
print(marg_co.summary())
```
