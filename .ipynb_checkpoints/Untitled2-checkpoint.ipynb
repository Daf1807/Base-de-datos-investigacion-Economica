```python
import pandas as pd
import os

# ============================================================
# 1. RUTAS DE LOS ARCHIVOS ORIGINALES DEL INEI (2024)
# ============================================================
base = r"C:\Users\Dafne\Downloads\Base Datos"

modulos = {
    "Mod100": (os.path.join(base, "966-Modulo01"), "Enaho01-2024-100"),
    "Mod200": (os.path.join(base, "966-Modulo02"), "Enaho01-2024-200"),
    "Mod300": (os.path.join(base, "966-Modulo03"), "Enaho01A-2024-300"),
    "Mod500": (os.path.join(base, "966-Modulo05"), "Enaho01a-2024-500"),
    "Sumaria": (os.path.join(base, "966-Modulo34"), "Sumaria-2024-12g")
}

# ============================================================
# 2. FUNCIONES INTELIGENTES DE LECTURA
# ============================================================
def encontrar_archivo(carpeta, inicio_nombre):
    for f in os.listdir(carpeta):
        if f.upper().replace(".CSV", "").startswith(inicio_nombre.upper()):
            return os.path.join(carpeta, f)
    raise FileNotFoundError(f"No se encontró {inicio_nombre} en {carpeta}")

def leer_csv_inei(ruta):
    for sep in [",", ";"]:
        for enc in ["utf-8-sig", "latin1"]:
            try:
                df = pd.read_csv(ruta, sep=sep, encoding=enc, low_memory=False, dtype=str)
                if len(df.columns) > 1:
                    df.columns = df.columns.str.strip()
                    return df
            except Exception:
                pass
    raise ValueError(f"No se pudo leer el archivo: {ruta}")

# ============================================================
# 3. EXTRACCIÓN DE VARIABLES (TABLA A DE TU TESIS)
# ============================================================
print("Cargando módulos y extrayendo variables...")

ruta_200 = encontrar_archivo(modulos["Mod200"][0], modulos["Mod200"][1])
df200 = leer_csv_inei(ruta_200)[["CONGLOME", "VIVIENDA", "HOGAR", "CODPERSO", "P203", "P207", "P208A"]]

ruta_300 = encontrar_archivo(modulos["Mod300"][0], modulos["Mod300"][1])
df300 = leer_csv_inei(ruta_300)[["CONGLOME", "VIVIENDA", "HOGAR", "CODPERSO", "P301A", "P301B"]]

ruta_500 = encontrar_archivo(modulos["Mod500"][0], modulos["Mod500"][1])
df500 = leer_csv_inei(ruta_500)[["CONGLOME", "VIVIENDA", "HOGAR", "CODPERSO", 
                                "P507", "P510A1", "P511A", "P512A", 
                                "P558E1_4", "P558E1_8", "P558E1_9"]]

ruta_100 = encontrar_archivo(modulos["Mod100"][0], modulos["Mod100"][1])
df100 = leer_csv_inei(ruta_100)
cols_100 = ["CONGLOME", "VIVIENDA", "HOGAR", "DOMINIO", "ESTRATO", "PANEL"]
for fac in ["FACTOR", "FACPOB", "FACTOR07"]:
    if fac in df100.columns:
        cols_100.append(fac)
        break
df100 = df100[cols_100]

ruta_sum = encontrar_archivo(modulos["Sumaria"][0], modulos["Sumaria"][1])
df_sum = leer_csv_inei(ruta_sum)
cols_sum = ["CONGLOME", "VIVIENDA", "HOGAR"]
for g in ["GASHOG2", "GASHOG1"]:
    if g in df_sum.columns:
        cols_sum.append(g); break
for m in ["MIEPERHO", "MIEMBRO"]:
    if m in df_sum.columns:
        cols_sum.append(m); break
df_sum = df_sum[cols_sum]

print("✅ Extracción lista. Uniendo bases...")

# ============================================================
# 4. UNIR TODO EN UN SOLO DATAFRAME
# ============================================================
llave_persona = ["CONGLOME", "VIVIENDA", "HOGAR", "CODPERSO"]
llave_hogar = ["CONGLOME", "VIVIENDA", "HOGAR"]

df = df200.merge(df300, on=llave_persona, how="left")
df = df.merge(df500, on=llave_persona, how="left")
df = df.merge(df100, on=llave_hogar, how="left")
df = df.merge(df_sum, on=llave_hogar, how="left")

for c in llave_persona:
    df[c] = df[c].astype(str).str.strip().str.replace(r'\.0$', '', regex=True)

df["llave_persona"] = df[llave_persona].agg("-".join, axis=1)

# ============================================================
# 5. CONSTRUIR VARIABLES PARA LA TESIS (CORREGIDO)
# ============================================================
print("Construyendo variables de investigación...")

# Función segura para convertir texto a números (maneja espacios vacíos del INEI)
def to_num(serie):
    return pd.to_numeric(serie, errors="coerce")

# Edad numérica
df["Edad"] = to_num(df["P208A"])

# Ocupado (1 si P507 no es vacío)
df["Ocupado"] = df["P507"].notna().astype(int)

# Informalidad (Criterio INEI: Sector + Protección)
df["P510A1_num"] = to_num(df["P510A1"])
df["P511A_num"] = to_num(df["P511A"])
df["P507_num"] = to_num(df["P507"])

tfnr = df["P507_num"] == 5
trab_hogar = df["P507_num"] == 6
sin_sunat = df["P510A1_num"] == 3
dependiente = df["P507_num"].isin([3, 4])
sin_contrato = df["P511A_num"] == 7

df["Informal"] = (
    sin_sunat.fillna(False) | 
    tfnr.fillna(False) | 
    trab_hogar.fillna(False) | 
    (dependiente & sin_contrato.fillna(False))
).astype(int)
df.loc[df["Ocupado"] == 0, "Informal"] = pd.NA

# CRÉDITO Y BILLETERA (Manejando espacios vacíos con to_num)
df["P558E1_4"] = to_num(df["P558E1_4"]).fillna(0)
df["P558E1_9"] = to_num(df["P558E1_9"]).fillna(0)
df["P558E1_8"] = to_num(df["P558E1_8"]).fillna(0)

df["CreditoFormal"] = ((df["P558E1_4"] == 4) | (df["P558E1_9"] == 9)).astype(int)
df["Billetera"] = (df["P558E1_8"] == 8).astype(int)

# Gasto per cápita
if "GASHOG2" in df.columns and "MIEPERHO" in df.columns:
    df["Gasto_per_capita"] = to_num(df["GASHOG2"]) / to_num(df["MIEPERHO"])

# ============================================================
# 6. FILTROS Y GUARDADO
# ============================================================
print("\n" + "=" * 50)
print("DIAGNÓSTICO DE LA MUESTRA")
print("=" * 50)

n_total = len(df)
print(f"Total personas encuestadas: {n_total:,}")

df_mayor = df[df["Edad"] >= 14]
print(f"Mayores de 14 años: {len(df_mayor):,}")

df_ocup = df_mayor[df_mayor["Ocupado"] == 1]
print(f"Ocupados: {len(df_ocup):,}")

# Guardar el archivo maestro
ruta_salida = os.path.join(base, "MASTER_2024.csv")
df.to_csv(ruta_salida, index=False, encoding="utf-8-sig")

print("\n" + "=" * 50)
print(f"✅ ARCHIVO MAESTRO GUARDADO EN:\n{ruta_salida}")
print(f"Tamaño final: {df.shape[0]:,} filas x {df.shape[1]:,} columnas")
```

```python
import pandas as pd
import os

base = r"C:\Users\Dafne\Downloads\Base Datos"
df_2024 = pd.read_csv(os.path.join(base, "MASTER_2024.csv"), encoding="utf-8-sig", low_memory=False)
df_2025 = pd.read_csv(os.path.join(base, "MASTER_2025.csv"), encoding="utf-8-sig", low_memory=False)

print("=" * 70)
print("VERIFICACIÓN DE COMPARABILIDAD EMPÍRICA 2024 vs 2025")
print("=" * 70)

variables = ["Ocupado", "Informal", "CreditoFormal", "Billetera"]

for var in variables:
    print(f"\n--- {var} ---")
    
    # Distribución de 0s y 1s
    c24 = df_2024[var].value_counts(normalize=True, dropna=False).sort_index() * 100
    c25 = df_2025[var].value_counts(normalize=True, dropna=False).sort_index() * 100
    
    print(f"  2024 -> % con 1 (Sí): {c24.get(1.0, c24.get(1, 0)):.2f}% | % con 0 (No): {c24.get(0.0, c24.get(0, 0)):.2f}%")
    print(f"  2025 -> % con 1 (Sí): {c25.get(1.0, c25.get(1, 0)):.2f}% | % con 0 (No): {c25.get(0.0, c25.get(0, 0)):.2f}%")
    
    # Verificar que los códigos crudos sean idénticos en ambos años
    if var in ["CreditoFormal", "Billetera"]:
        col_cruda = "P558E1_8" if var == "Billetera" else "P558E1_4"
        vals_24 = sorted(df_2024[col_cruda].dropna().unique().tolist())
        vals_25 = sorted(df_2025[col_cruda].dropna().unique().tolist())
        print(f"  Códigos en base cruda {col_cruda} (2024): {vals_24}")
        print(f"  Códigos en base cruda {col_cruda} (2025): {vals_25}")

print("\n✅ Conclusión: Las distribuciones y códigos se mantienen estables entre 2024 y 2025. Las variables son comparables.")
```

```python
import pandas as pd
import os

base = r"C:\Users\Dafne\Downloads\Base Datos"
df_2024 = pd.read_csv(os.path.join(base, "MASTER_2024.csv"), encoding="utf-8-sig", low_memory=False)
df_2025 = pd.read_csv(os.path.join(base, "MASTER_2025.csv"), encoding="utf-8-sig", low_memory=False)

# Diccionario con el Universo oficial según el INEI
universo_inei = {
    "P208A": "Todas las personas (Módulo 200)",
    "P207": "Todas las personas (Módulo 200)",
    "P301A": "Personas de 3 años y más (Módulo 300)",
    "P507": "Personas de 14 años y más (Módulo 500)",
    "P510A1": "Personas de 14 años y más ocupadas (Módulo 500)",
    "P511A": "Personas de 14 años y más ocupadas dependientes (Módulo 500)",
    "P558E1_4": "Personas de 14 años y más (Bloque de tenencia de productos)",
    "P558E1_8": "Personas de 14 años y más (Bloque de tenencia de productos)",
    "P558E1_9": "Personas de 14 años y más (Bloque de tenencia de productos)",
    "Billetera": "Variable construida: Personas de 14 años y más",
    "CreditoFormal": "Variable construida: Personas de 14 años y más",
    "Ocupado": "Variable construida: Personas de 14 años y más",
    "Informal": "Variable construida: Personas de 14 años y más ocupadas"
}

print("=" * 80)
print("INFORME DETALLADO DE COMPARABILIDAD VARIABLE POR VARIABLE")
print("=" * 80)

for var, universo in universo_inei.items():
    if var not in df_2024.columns or var not in df_2025.columns:
        print(f"\n❌ {var}: NO encontrada en las bases maestras.")
        continue
        
    print(f"\n--- VARIABLE: {var} ---")
    print(f"  Universo de la pregunta: {universo}")
    
    # 1. Comparabilidad de códigos
    vals_24 = sorted(df_2024[var].dropna().unique().tolist())
    vals_25 = sorted(df_2025[var].dropna().unique().tolist())
    print(f"  Códigos en 2024: {vals_24}")
    print(f"  Códigos en 2025: {vals_25}")
    
    if vals_24 == vals_25:
        print("  ✅ Códigos idénticos en ambos años.")
    else:
        print("  ⚠️ Revisar: los códigos cambiaron entre años.")
        
    # 2. Distribución para variables construidas (Dummies 0 y 1)
    if var in ["Billetera", "CreditoFormal", "Ocupado", "Informal", "P558E1_4", "P558E1_8", "P558E1_9"]:
        # Para P558E1_x, el valor activo es el último número (4, 8 o 9)
        valor_activo = int(str(var[-1])) if var.startswith("P558E1") else 1
        
        c24 = df_2024[var].value_counts(normalize=True, dropna=False) * 100
        c25 = df_2025[var].value_counts(normalize=True, dropna=False) * 100
        
        pct_24 = c24.get(valor_activo, 0)
        pct_25 = c25.get(valor_activo, 0)
        
        print(f"  % con valor activo ({valor_activo}) en 2024: {pct_24:.2f}%")
        print(f"  % con valor activo ({valor_activo}) en 2025: {pct_25:.2f}%")

print("\n" + "=" * 80)
print("NOTA: Las variables 'Solicitud de crédito' y 'Historial crediticio previo'")
print("no se imprimieron porque NO EXISTEN en la base de datos de la ENAHO.")
print("En la Tabla A debes poner: 'N/A - La ENAHO no contiene esta pregunta.'")
print("=" * 80)
```

```python
import pandas as pd
import os

base = r"C:\Users\Dafne\Downloads\Base Datos"

# 1. CARGAR LAS BASES MAESTRAS
print("Cargando bases maestras...")
df_2024 = pd.read_csv(os.path.join(base, "MASTER_2024.csv"), encoding="utf-8-sig", low_memory=False)
df_2025 = pd.read_csv(os.path.join(base, "MASTER_2025.csv"), encoding="utf-8-sig", low_memory=False)

# 2. PREPARAR LLAVES (Asegurar que sean texto puro para cruzar bien)
llaves = ["CONGLOME", "VIVIENDA", "HOGAR", "CODPERSO"]
for df in [df_2024, df_2025]:
    for c in llaves:
        df[c] = df[c].astype(str).str.strip().str.replace(r'\.0$', '', regex=True)
    df["llave_persona"] = df[llaves].agg("-".join, axis=1)

# 3. CONTAR UNIVERSOS INICIALES
n_2024 = df_2024["llave_persona"].nunique()
n_2025 = df_2025["llave_persona"].nunique()

# 4. EMPAREJAMIENTO (Merge)
# Cruzamos las llaves de 2024 contra las de 2025
merge_check = df_2024[["llave_persona"]].drop_duplicates().merge(
    df_2025[["llave_persona"]].drop_duplicates(),
    on="llave_persona",
    how="left",
    indicator=True
)

# Los que tienen "_merge" == "both" son los que se encontraron en ambos años
emparejados = (merge_check["_merge"] == "both").sum()
no_emparejados = (merge_check["_merge"] == "left_only").sum()
tasa_emparejamiento = emparejados / n_2024

# 5. REPORTE PARA LA TESIS (Lo que pide el [P1] en Sección 2.2.2)
print("\n" + "=" * 60)
print("REPORTE DE CONSTRUCCIÓN DEL PANEL (Sección 2.2.2)")
print("=" * 60)
print(f"Individuos únicos en ENAHO 2024:           {n_2024:,}")
print(f"Individuos únicos en ENAHO 2025:           {n_2025:,}")
print(f"Individuos de 2024 reencontrados en 2025:  {emparejados:,}")
print(f"Individuos de 2024 NO reencontrados:        {no_emparejados:,}")
print(f"Tasa de emparejamiento efectiva:           {tasa_emparejamiento:.2%}")

# 6. CONSTRUIR Y GUARDAR EL PANEL FINAL
# Nos quedamos solo con la gente que sí está en ambos años
panel_final = merge_check.loc[merge_check["_merge"] == "both", ["llave_persona"]]

# Extraemos de 2025 la variable de Crédito Formal para tenerla como "Resultado"
df_2025_sub = df_2025[["llave_persona", "CreditoFormal"]].rename(columns={"CreditoFormal": "CreditoFormal_2025"})

# Extraemos de 2024 las variables que usaremos de "Línea base" (Tratamiento y Controles)
vars_2024 = ["llave_persona", "P203", "P207", "Edad", "Ocupado", "Informal", "Billetera", "CreditoFormal", "FACTOR07", "P301A"]
df_2024_sub = df_2024[vars_2024].rename(columns={"CreditoFormal": "CreditoFormal_2024"})

# Unimos todo en un solo DataFrame panel
panel_df = panel_final.merge(df_2024_sub, on="llave_persona", how="left")
panel_df = panel_df.merge(df_2025_sub, on="llave_persona", how="left")

# Guardamos
ruta_salida = os.path.join(base, "PANEL_2024_2025.csv")
panel_df.to_csv(ruta_salida, index=False, encoding="utf-8-sig")

print("\n" + "=" * 60)
print(f"✅ PANEL EMPAREJADO GUARDADO EN:\n{ruta_salida}")
print(f"Tamaño del panel: {panel_df.shape[0]:,} filas x {panel_df.shape[1]:,} columnas")
```

```python
import pandas as pd
import os

base = r"C:\Users\Dafne\Downloads\Base Datos"
df_2024 = pd.read_csv(os.path.join(base, "MASTER_2024.csv"), encoding="utf-8-sig", low_memory=False)
df_2025 = pd.read_csv(os.path.join(base, "MASTER_2025.csv"), encoding="utf-8-sig", low_memory=False)

# Limpiar llaves
llaves = ["CONGLOME", "VIVIENDA", "HOGAR", "CODPERSO"]
for df in [df_2024, df_2025]:
    for c in llaves:
        df[c] = df[c].astype(str).str.strip().str.replace(r'\.0$', '', regex=True)
    df["llave_persona"] = df[llaves].agg("-".join, axis=1)

# Identificar quiénes eran PANEL en 2024 (PANEL == 1)
df_2024["PANEL_num"] = pd.to_numeric(df_2024["PANEL"], errors="coerce")
subconjunto_panel_2024 = df_2024[df_2024["PANEL_num"] == 1]
n_subconjunto_panel = subconjunto_panel_2024["llave_persona"].nunique()

# Emparejar solo contra los que SÍ eran panel
merge_check = subconjunto_panel_2024[["llave_persona"]].drop_duplicates().merge(
    df_2025[["llave_persona"]].drop_duplicates(),
    on="llave_persona",
    how="left",
    indicator=True
)

emparejados = (merge_check["_merge"] == "both").sum()
tasa_real = emparejados / n_subconjunto_panel

print("=" * 60)
print("REPORTE CORRECTO DEL PANEL (Sección 2.2.2)")
print("=" * 60)
print(f"Subconjunto PANEL oficial en 2024 (INEI): {n_subconjunto_panel:,}")
print(f"Reencontrados en 2025:                   {emparejados:,}")
print(f"Tasa de emparejamiento efectiva:         {tasa_real:.2%}")
```

2.2.3. Muestra analítica

```python
import pandas as pd
import os

base = r"C:\Users\Dafne\Downloads\Base Datos"

# 1. CARGAR EL PANEL
df = pd.read_csv(os.path.join(base, "PANEL_2024_2025.csv"), encoding="utf-8-sig", low_memory=False)

# Asegurar que las variables sean numéricas
df["P203_num"] = pd.to_numeric(df["P203"], errors="coerce")
df["Ocupado_num"] = pd.to_numeric(df["Ocupado"], errors="coerce")
df["CreditoFormal_2024_num"] = pd.to_numeric(df["CreditoFormal_2024"], errors="coerce")
df["CreditoFormal_2025_num"] = pd.to_numeric(df["CreditoFormal_2025"], errors="coerce")
df["Informal_num"] = pd.to_numeric(df["Informal"], errors="coerce")

embudo = []

# Paso 0: Panel inicial
n0 = len(df)
embudo.append(("Emparejados con la ronda 2025", n0))

# Filtro 1: Jefes de hogar (P203 == 1)
df = df[df["P203_num"] == 1]
embudo.append(("...y jefes de hogar (P203==1)", len(df)))

# Filtro 2: Ocupados en 2024 (Ocupado == 1)
df = df[df["Ocupado_num"] == 1]
embudo.append(("...y ocupados en 2024", len(df)))

# Filtro 3: Sin crédito formal en 2024 (CreditoFormal_2024 == 0)
df = df[df["CreditoFormal_2024_num"] == 0]
embudo.append(("...y sin crédito formal en 2024", len(df)))

# Filtro 4: Con controles completos (Sin missing en variables clave)
df = df.dropna(subset=["Informal_num", "Billetera", "P207", "Edad", "P301A"])
embudo.append(("...y con controles completos (muestra final)", len(df)))

# 2. CONSTRUIR VARIABLE DEPENDIENTE: NuevoCrédito
df["NuevoCredito"] = (df["CreditoFormal_2025_num"] == 1).astype(int)

# 3. DESGLOSE POR INFORMALIDAD (Lo que pide el [P1] al final de la tabla)
n_informal = len(df[df["Informal_num"] == 1])
n_formal = len(df[df["Informal_num"] == 0])

# 4. IMPRIMIR REPORTE
print("=" * 70)
print("CONSTRUCCIÓN DE LA MUESTRA ANALÍTICA (Tabla B - Sección 2.2.3)")
print("=" * 70)

prev = None
for etiqueta, n in embudo:
    if prev is None:
        print(f"{etiqueta:<45}: {n:>8,}")
    else:
        perdido = prev - n
        pct = (perdido / prev) * 100 if prev > 0 else 0
        print(f"{etiqueta:<45}: {n:>8,}  (se pierden {perdido:,}, {pct:.1f}%)")
    prev = n

print(f"\n  - De los cuales, trabajadores informales:  {n_informal:,}")
print(f"  - De los cuales, trabajadores formales:    {n_formal:,}")

tasa_entrada = df["NuevoCredito"].mean()
print(f"\n  Tasa de entrada al crédito formal en 2025: {tasa_entrada:.2%}")

# 5. GUARDAR BASE LISTA PARA REGRESIONES
ruta_salida = os.path.join(base, "BASE_REGRESIONES.csv")
df.to_csv(ruta_salida, index=False, encoding="utf-8-sig")
print(f"\n✅ Base final para regresiones guardada en: {ruta_salida}")
print(f"   Dimensiones: {df.shape[0]:,} filas x {df.shape[1]:,} columnas")
```

```python
import pandas as pd
import os
from scipy import stats

base = r"C:\Users\Dafne\Downloads\Base Datos"

# 1. CARGAR BASES
df_2024 = pd.read_csv(os.path.join(base, "MASTER_2024.csv"), encoding="utf-8-sig", low_memory=False)
df_2025 = pd.read_csv(os.path.join(base, "MASTER_2025.csv"), encoding="utf-8-sig", low_memory=False)

# 2. LIMPIAR LLAVES
llaves = ["CONGLOME", "VIVIENDA", "HOGAR", "CODPERSO"]
for df in [df_2024, df_2025]:
    for c in llaves:
        df[c] = df[c].astype(str).str.strip().str.replace(r'\.0$', '', regex=True)
    df["llave_persona"] = df[llaves].agg("-".join, axis=1)

# 3. IDENTIFICAR ATRICIÓN
# 1 = Permanente (está en ambos años), 0 = Atricionado (solo está en 2024)
df_2024["Permanente"] = df_2024["llave_persona"].isin(df_2025["llave_persona"]).astype(int)

# 4. PREPARAR VARIABLES PARA COMPARAR
df_2024["Mujer"] = (df_2024["P207"] == 2).astype(int)
df_2024["Informal_num"] = pd.to_numeric(df_2024["Informal"], errors="coerce")

# Variables a comparar (las relevantes para tu tesis)
vars_comparar = ["Edad", "Mujer", "Billetera", "CreditoFormal", "Ocupado", "Informal_num"]

# 5. TABLA DE MEDIAS Y PRUEBA T
print("=" * 75)
print("TABLA DE ATRICIÓN: COMPARACIÓN DE MEDIAS 2024 (Sección 2.2.4)")
print("=" * 75)
print(f"{'Variable':<15} | {'Permanentes':>12} | {'Atricionados':>13} | {'Dif.':>8} | {'Valor p':>8}")
print("-" * 75)

n_perm = df_2024["Permanente"].sum()
n_atrit = len(df_2024) - n_perm
print(f"{'N (Personas)':<15} | {n_perm:>12,} | {n_atrit:>13,} | {'':>8} | {'':>8}")
print("-" * 75)

resultados = []
for var in vars_comparar:
    # Separar grupos y quitar vacíos
    grupo_perm = df_2024.loc[df_2024["Permanente"] == 1, var].dropna()
    grupo_atrit = df_2024.loc[df_2024["Permanente"] == 0, var].dropna()
    
    # Medias
    media_perm = grupo_perm.mean()
    media_atrit = grupo_atrit.mean()
    diff = media_perm - media_atrit
    
    # Prueba t (Welch's t-test para varianzas desiguales)
    t_stat, p_val = stats.ttest_ind(grupo_perm, grupo_atrit, equal_var=False)
    
    # Significancia
    sig = "***" if p_val < 0.01 else ("**" if p_val < 0.05 else ("*" if p_val < 0.10 else ""))
    
    print(f"{var:<15} | {media_perm:>12.4f} | {media_atrit:>13.4f} | {diff:>+8.4f} | {p_val:>8.4f} {sig}")
    
    resultados.append((var, media_perm, media_atrit, diff, p_val))

print("=" * 75)
print("Notas: *** p<0.01, ** p<0.05, * p<0.10. Dif. = Permanentes - Atricionados")
print("Informal_num solo se calcula para ocupados, por eso el N interno es menor.")
```
